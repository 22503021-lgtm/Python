```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path_2012 = r"Tabela5-sem_emprego_2012.csv"

t2012 = pd.read_csv(
    path_2012,
    sep=";",
    decimal=",",
    encoding="utf-8-sig"
)

t2012.head()
```

```python
path_2026 = r"Tabela5-sem_emprego_2026.csv"

t2026 = pd.read_csv(
    path_2026,
    sep=";",
    decimal=",",
    encoding="utf-8-sig"
)

t2026.head()
```

```python
comp = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)

comp.head()
```

```python
comp.info()
```

```python
comp["Diferença homens"] = (
    comp["Desocupados - homens (2026 T1)"] -
    comp["Desocupados - homens (2012 T1)"]
)

comp[[
    "Sigla",
    "Estado",
    "Desocupados - homens (2012 T1)",
    "Desocupados - homens (2026 T1)",
    "Diferença homens"
]]
```

```python
comp["Diferença mulheres"] = (
    comp["Desocupados - mulheres (2026 T1)"] -
    comp["Desocupados - mulheres (2012 T1)"]
)

comp[[
    "Sigla",
    "Estado",
    "Desocupados - mulheres (2012 T1)",
    "Desocupados - mulheres (2026 T1)",
    "Diferença mulheres"
]]
```

```python
comp.sort_values(
    "Diferença homens",
    ascending=False
)[[
    "Sigla",
    "Estado",
    "Diferença homens"
]]
```

```python
comp.sort_values(
    "Diferença mulheres",
    ascending=False
)[[
    "Sigla",
    "Estado",
    "Diferença mulheres"
]]
```
